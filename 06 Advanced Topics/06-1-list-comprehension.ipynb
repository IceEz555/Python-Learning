## 1. List Comprehension คืออะไร?
ชอร์ตคัต (Shortcut) ของการสร้าง List จาก List อื่นให้อยู่ในบรรทัดเดียว เป็น "เคล็ดลับความโปร" ที่คนใช้ Python บ่อยๆ ชอบมาก

```python
# -- แบบดั้งเดิม (For Loop ธรรมดา) --
numbers = [1, 2, 3, 4, 5]
squared_normal = []
for n in numbers:
    squared_normal.append(n * n)
print("แบบปกติ:", squared_normal)

# -- แบบ List Comprehension --
squared_comp = [n * n for n in numbers]
print("แบบ Comprehension:", squared_comp)
```
